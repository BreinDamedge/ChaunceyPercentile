```python
# read in the english dictionary
words:list[str] = []
with open("words_alpha.txt") as f:
    for line in f.readlines():
        words.append(line[:-1])

print(words[:10])
print(len(words), "words in dictionary")
```

```python
# convert to vowel percentage and sort
"""calculate the vowel percentage of a string"""
def vp(w_:str) -> float:
    vowels:set[str] = set([v for v in"aeiouy"])
    vct:int = 0
    for c in w_: 
        if c in vowels: vct += 1
    return vct/len(w_)

vowel_percentages:list[float] = sorted([vp(w) for w in words])
```

```python
# display the histogram
from matplotlib import pyplot as plt
plt.hist(vowel_percentages, bins=20)
plt.title("Vowel Percentage Of Words In The English Dictionary")
plt.xlabel("Percentage")
plt.ylabel("Count")
plt.show()
```

```python
def bsi(v_:float, cont_:list[float]) -> int:
    """binary search the container and return the index where v_ belongs"""
    mini, maxi = 0, len(cont_)-1
    at:int = int((maxi-mini)/2)+mini
    while True:
        if v_ < cont_[at]:
            maxi = at
        elif v_ > cont_[at]:
            mini = at
        else:
            return at
        at:int = int((maxi-mini)/2)+mini
        if mini == maxi: return mini

        


def percentile(word_:str, dist_:list[float]) -> float:
    wvp:float = vp(word_)
    # binary search the list till you find the place that wvp belongs
    i = bsi(wvp, dist_)
    return i/len(dist_)


def print_percentile(word_:str, dist_:list[float]=vowel_percentages) -> None:
    print(f'"{word_}" -> {percentile(word_, dist_)*100:.3}')
```

```python
print_percentile("chauncey")
print_percentile("gunther")
print_percentile("stuart")
print_percentile("beamer")
print_percentile("jacob")
```
